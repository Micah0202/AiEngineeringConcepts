```python
from dotenv import load_dotenv

load_dotenv()
from langchain_huggingface import HuggingFacePipeline
from langchain_core.prompts import PromptTemplate , ChatPromptTemplate
```

```python

llm = HuggingFacePipeline.from_model_id(
    model_id='Qwen/Qwen2.5-1.5B-Instruct',
    task='text-generation',
    pipeline_kwargs={"max_new_tokens":2048}
    
)
```

```python
algo_tutor_template = """
You are an expert computer Science engineer . Explain the following algorithm: {algorithm_name}

Follow the EXACT structure : 

### 1. The concept (What it is  ? )
Explain what problem this algorithm solves in 2 sentences .

###2 . The efficiency (Complexity analysis)
Define the time complexity and space complexity of the {algorithm_name}

###3. Sample trace run
Show  a simple dry-run using the sample input: {sample_input}

Answer ::

"""
```

```python
prompt = PromptTemplate(
    template=algo_tutor_template,
    input_variables=["algorithm_name","sample_input"]
)
```

```python
final_prompt = prompt.format(algorithm_name="Bubble  Sort" ,  sample_input="[5,1,4,2]")
```

```python
print(final_prompt)
```

```python
chain = prompt | llm  #pipe operator  takes the prompt and feeds it to llm 

result = chain.invoke({
    "algorithm_name": "Binary search",
     "sample_input" : "Target 7 in List [1,3,5,7,9,11,13,15,16]"
})
```

```python
print(result)
```
