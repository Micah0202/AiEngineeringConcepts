```python
from langchain.agents import create_agent 
from langchain_core.tools import tool 
```

```python
@tool
def get_weather(location: str) -> str:
    """Get the weather in a given location"""
    return f"The weather in {location} is sunny and 25 degrees Celsius"
```

```python
agent = create_agent(
    model="openai:gpt-4o-mini",
    tools=[get_weather],
    system_prompt="You are a weather agent. You are given a location and you need to return the weather in that location."
)
```

```python
result = agent.invoke({
    "messages": [
        {"role": "user", "content": "What is the weather in Tokyo compared to London ?"}
    ]
})
```
